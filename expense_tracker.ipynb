```python
expenses = []

while(True):
  print("-------------Expense Tracker-----------")
  print("1.Add expenses")
  print("2.View expenses")
  print("3.Total expenses")
  print("4.Exit")

  choice= int(input(" Enter your choice : "))

  if choice==1:
    name = input(" Enter expenses name : ")
    amount = float(input(" Enter expenses amount : "))
    expenses.append([name,amount])
    print(" Expenses added succesfully !")

  elif choice == 2:
      print(" Your expenses : ")
      for item in expenses :
        print(f"{item[0]} : Rs {item[1]}")

  elif choice == 3:
      total = 0
      for item in expenses :
        total += item[1]

      print("Total expenses :  Rs " , total)

  elif choice== 4:
      print(" Thank you ! you are exiting expense tracker")
      break

  else :
      print("Invalid choice ! please try again ")
```
